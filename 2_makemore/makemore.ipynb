## Karpathy's MakeMore: bigram character-level language model

### First sight at names.txt

```python
words = open('names.txt', 'r').read().splitlines()
print(f'Number of names: {len(words)}')
print(f'Len of shortest names: {min(len(w) for w in words)}')
print(f'Len of longest names: {max(len(w) for w in words)}')

print(f'Example of names: {words[:5]}')

chars = sorted(list(set(''.join(words))))
print(f'Characters into names: {chars} [len = {len(chars)}]')
print(f'{len(chars) = }]')
```

Create bigrams and count how often they appear

This process requires special char for start and end of a name. 
Intuitively you would use something like "-S-" and "-E-". 

Since the char for start cannot be the second element of the bigram and similarly the char for end cannot be first, one single special char is enough. We'll use the dot ".", both for start and for end.

Also, strings has to be converted to numbers. The easiest way is casting to int, but we need to create a mapper (stoi: string to int) and one inverter (itos: int to string).

```python
chars = sorted(list(set(''.join(words))))
stoi = {s:i+1 for i,s in enumerate(chars)}
stoi['.'] = 0 # special char
itos = {i:s for s,i in stoi.items()}
print(f'Example of mapping in stoi: {[(k,v) for (k,v) in stoi.items()][:5]}')
print(f'Example of mapping in itos: {[(k,v) for (k,v) in itos.items()][:5]}')
b = {}
for w in words:
  chs = ['.'] + list(w) + ['.']
  for ch1, ch2 in zip(chs, chs[1:]):
    bigram = (ch1, ch2)
    b[bigram] = b.get(bigram, 0) + 1
print(f'Most common bigrams: \n{sorted(b.items(), key = lambda kv: -kv[1])[:5]}')
```
